```python
# Reference: https://huggingface.co/learn/llm-course/chapter11/1
# [|system|] [|endofturn|]
# [|human|] ... [|endofturn|]
# [|assistant|] ...
```

```python
from transformers import AutoTokenizer

messages = [
    {"role": "system", "content": "You are a helpful assistant."},
    {"role": "user", "content": "How are you?"},
    {"role": "assistant", "content": "I'm fine, thank you and you?"},
]
exaone_tokenizer = AutoTokenizer.from_pretrained("LGAI-EXAONE/EXAONE-3.5-2.4B-Instruct")
exaone_chat = exaone_tokenizer.apply_chat_template(messages, tokenize=False)
print(exaone_chat)
```
